```python
import pandas as pd
```

```python
df = pd.read_csv("train.csv")
df
```

```python
df.info()
```

```python
df["Age"]= df["Age"].fillna(df["Age"].median())
df["Age"]
df["Cabin"]=df["Cabin"].fillna(df["Cabin"].mode()[0])
df["Cabin"]
df["Embarked"]= df["Embarked"].fillna(df["Cabin"].mode()[0])
df["Embarked"]
```

```python
df.info()
```

```python
df['Age']=df["Age"].astype(int)
df["Age"]
df.info()
```

```python
df.rename(columns={'Pclass': 'passengerclass'}, inplace=True)
df.info()
df.rename(columns = {'SibSp':'siblings/sponses'}, inplace = True)
df.info()
df.rename(columns = {"Parch": "number of parents and children"}, inplace = True)

df.info()
```

```python
# df.info()
# df["Embarked"].unique()
df['Embarked'] = df['Embarked'].replace('B96 B98', 'S')
df["Embarked"]= df["Embarked"].replace('B96 B98','S')
df["Embarked"].unique()
df.describe()
# object (text) टाइप भएका कोलमहरू मात्र छानेर तिनीहरूको unique भ्यालु हेर्ने:
for col in df.select_dtypes(include='object').columns:
    print(f"--- Unique values in {col} ---")
    print(df[col].unique())
    print("\n")
```
